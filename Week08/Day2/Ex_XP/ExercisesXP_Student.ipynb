## 1. Problem Statement

The objective of this project is to develop a supervised machine learning classification model to predict the probability of loan default. The target variable, `loan_default`, will indicate whether a borrower defaults (`1`) or does not default (`0`) within 12 months of loan origination, based on a predefined default criterion.

The model will support financial institutions in assessing credit risk, reducing potential financial losses and making more informed lending decisions.

## 2. Data Types and Their Relevance

To develop an effective predictive model, the following types of data should be collected:

- **Applicant Information:** Income, employment status and employment length can help assess a borrower's financial stability and repayment capacity.
- **Loan Information:** Loan amount, interest rate, repayment term, monthly installments and loan purpose can help identify how loan characteristics relate to default risk.
- **Credit History:** Credit scores, previous defaults and late payments can provide insights into the borrower's past repayment behavior.
- **Financial Situation:** Existing debts, monthly financial obligations and the debt-to-income (DTI) ratio can help evaluate the borrower's overall financial burden.
- **Economic Indicators:** Unemployment rates and other relevant economic conditions can provide additional context for understanding default risk.

## 3. Data Collection Sources

Data can be collected and integrated from several sources:

- **Internal Bank Records:** Historical loan applications, loan agreements and repayment records can be extracted from the institution's internal databases. These records will also provide the historical loan outcomes required to define the target variable.
- **Credit Bureaus:** Credit scores, outstanding debts and repayment histories can be obtained from authorized credit reporting agencies, subject to applicable access requirements.
- **Loan Application Forms:** Income, employment information, financial obligations and loan purpose can be collected directly from applicants and verified using appropriate supporting documents.
- **Official Economic Databases:** Macroeconomic indicators can be obtained from central banks or national statistical agencies and matched to the relevant application dates.

The collected datasets should be integrated using secure borrower and loan identifiers, while ensuring that each predictive feature was available at the time of the lending decision.

## 4. Risks and Constraints

Several risks must be considered throughout the project. **Privacy and regulatory compliance** require protecting sensitive financial information, restricting access and respecting applicable data protection and lending regulations. **Data quality** must be assessed by identifying missing values, inconsistencies and inaccurate records. **Sampling bias** may arise because historical repayment outcomes are generally available only for approved loans, meaning the training dataset may not represent all applicants. Fairness must also be evaluated to reduce the risk of discriminatory outcomes. Additionally, **data leakage** must be prevented by excluding information that becomes available only after the lending decision. Finally, appropriate **data governance** is necessary to establish clear responsibilities for data access, security, retention and quality.

```python

# This piece of code is already prefilled, run it to execute it and see the results.
# It demonstrates standard classification metrics for binary loan default prediction.

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, average_precision_score, confusion_matrix, classification_report

# Please replace these placeholders with your true labels and predicted probabilities.
y_true = [0,1,0,1,0,0,1,0,1,0]            # placeholder labels
y_pred_proba = [0.05,0.80,0.10,0.65,0.20,0.15,0.70,0.30,0.85,0.25]  # placeholder probabilities

# You should set a decision threshold that reflects the precision–recall trade-off for your business case.
threshold = 0.5
y_pred = [1 if p >= threshold else 0 for p in y_pred_proba]

print("Accuracy:", round(accuracy_score(y_true, y_pred), 4))
print("Precision:", round(precision_score(y_true, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_true, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_true, y_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_true, y_pred_proba), 4))
print("PR-AUC (Average Precision):", round(average_precision_score(y_true, y_pred_proba), 4))
print("\nConfusion matrix:\n", confusion_matrix(y_true, y_pred))
print("\nClassification report:\n", classification_report(y_true, y_pred, zero_division=0))
```
