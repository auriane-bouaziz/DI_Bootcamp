🌟 Exercise 5 : Analyzing US-CERT Security Alerts
Instructions
Write a Python program to get the number of security alerts issued by US-CERT in the current year.
Source

```python
import requests
from bs4 import BeautifulSoup
from datetime import datetime

# Define the target URL
url = "https://www.cisa.gov/news-events/cybersecurity-advisories"

# Define filters
params = {
    "f[0]": "advisory_type:93",
    "f[1]": f"release_date_year:{datetime.now().year}"
}

# Fetch the webpage
response = requests.get(url, params=params, timeout=15)

print("Status code:", response.status_code)
print("URL:", response.url)
```

```python
import re

# Parse HTML
soup = BeautifulSoup(response.text, "html.parser")

# Extract page text
text = soup.get_text(" ", strip=True)

# Find the total number of results
match = re.search(
    r"Showing\s+[\d,]+\s+to\s+[\d,]+\s+of\s+([\d,]+)\s+results",
    text,
    re.IGNORECASE
)

if match:
    total_alerts = int(match.group(1).replace(",", ""))
    print(f"Total security alerts: {total_alerts}")
else:
    print("Result counter not found. Check the page structure and filters.")
```

### Exercise 5: Analyzing US-CERT Security Alerts

**Objective:** Retrieve and count the security alerts published by CISA during the current year (2026).

**Method:**

* Used `requests` to retrieve the CISA cybersecurity advisories webpage.
* Applied URL parameters to filter publications by alert type and year.
* Used BeautifulSoup and regular expressions to extract the number of results.
* Checked whether the requested filters were correctly applied.

**Result:**

The script successfully retrieved the webpage and extracted a total of 5,078 results. However, further validation showed that CISA had not applied the requested filters.

# Exercise 6: Scraping Movie Details

## Objective

Extract the movie title, release year, and a brief summary
for 10 movies from an IMDb list using BeautifulSoup.

## Methodology

The original IMDb list provided in the exercise was private
and could not be accessed.

I therefore selected an alternative public IMDb list
containing 10 movies released in 2020.

Since IMDb restricted automated access from Google Colab,
I downloaded the HTML page manually and imported it into
the notebook.

Using BeautifulSoup, I inspected the HTML structure,
identified the relevant movie containers, and extracted
the required information.

Finally, I organized the results into a Pandas DataFrame
and checked for missing values.

Note: The alternative list contains 10 predefined movies,
rather than a random selection from the original list.

```python
# Import required libraries
from google.colab import files
from bs4 import BeautifulSoup
import pandas as pd
import re
```

```python
from google.colab import files

uploaded = files.upload()

html = next(iter(uploaded.values()))
soup = BeautifulSoup(html, "html.parser")
```

```python
# Exercise 6: Extract movie details

# Select movie title links
movie_titles = soup.select('a[href*="ls_t_"]')

movies_data = []

for movie_link in movie_titles:

    # 1. Extract movie title
    title = re.sub(
        r"^\d+\.\s*",
        "",
        movie_link.get_text(strip=True)
    )

    # 2. Extract release year
    metadata = list(movie_link.parents)[3].get_text(" ", strip=True)

    year_match = re.search(r"\b(?:19|20)\d{2}\b", metadata)

    year = int(year_match.group()) if year_match else None

    # 3. Find the movie container
    movie = movie_link.find_parent(
        "li",
        class_="ipc-metadata-list-summary-item"
    )

    # 4. Extract movie summary
    summary_element = movie.select_one(
        ".ipc-html-content-inner-div"
    )

    summary = (
        summary_element.get_text(" ", strip=True)
        if summary_element else None
    )

    # 5. Store extracted information
    movies_data.append({
        "Movie": title,
        "Year": year,
        "Summary": summary
    })

# Create the final DataFrame
df_movies = pd.DataFrame(movies_data)

display(df_movies)
```

```python
print("Number of movies:", len(df_movies))

print("\nMissing values:")
print(df_movies.isna().sum())
```
