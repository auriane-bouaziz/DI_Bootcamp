```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# Define the URL
url = "https://github.com/topics"

# Send HTTP request
response = requests.get(url, timeout=15)

# Check status code
print("Status code:", response.status_code)

# Display the first 100 characters
print("\nHTML Preview:")
print(response.text[:100])
```

```python
# Save HTML content
if response.status_code == 200:

    with open("webpage.html", "w", encoding="utf-8") as file:
        file.write(response.text)

    print("HTML successfully saved!")

else:
    print("Unable to save the HTML.")
```

```python
# Read saved HTML file
with open("webpage.html", "r", encoding="utf-8") as file:
    html = file.read()

# Parse HTML with BeautifulSoup
soup = BeautifulSoup(html, "html.parser")

print("Page title:", soup.title.get_text(strip=True))
```

```python
# Inspect headings and their HTML classes
for heading in soup.find_all(["h2", "h3"])[:15]:
    print("TAG:", heading.name)
    print("TEXT:", heading.get_text(" ", strip=True))
    print("CLASS:", heading.get("class"))
    print("---")
```

```python
# Find potential topic cards
topic_links = soup.select('a[href^="/topics/"]')

# Keep links containing at least two paragraphs
topic_cards = [
    link for link in topic_links
    if len(link.find_all("p")) >= 2
]

print("Number of topic cards:", len(topic_cards))

# Inspect the first 3 topics
for card in topic_cards[:3]:
    print("\nURL:", card.get("href"))

    for paragraph in card.find_all("p"):
        print("TEXT:", paragraph.get_text(" ", strip=True))
```

```python
# Extract topic titles and descriptions

titles = []
descriptions = []

for card in topic_cards:
    paragraphs = card.find_all("p")

    titles.append(paragraphs[0].get_text(strip=True))
    descriptions.append(paragraphs[1].get_text(" ", strip=True))

# Verify the extraction
print("Number of titles:", len(titles))
print("Titles:", titles)

print("\nNumber of descriptions:", len(descriptions))
print("Descriptions:", descriptions)
```

```python
# Create a dictionary
topics_data = {
    "Title": titles,
    "Description": descriptions
}

# Convert dictionary into a DataFrame
df_topics = pd.DataFrame(topics_data)

# Display the DataFrame
display(df_topics)
```

```python
# Validate the extracted data

print("DataFrame shape:", df_topics.shape)

print("\nMissing values:")
print(df_topics.isna().sum())

print("\nDuplicate titles:")
print(df_topics["Title"].duplicated().sum())
```
