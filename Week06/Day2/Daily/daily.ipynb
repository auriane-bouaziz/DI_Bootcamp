```python
cities = [
    "City 1", "City 2", "City 3", "City 4", "City 5",
    "City 6", "City 7", "City 8", "City 9", "City 10"
]

months = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]
```

```python
len(cities)
```

```python
len(months)
```

```python
np.random.seed(42)

temperature_data = np.random.uniform(
    low=-5,
    high=35,
    size=(10, 12)
)

temperature_data = np.round(temperature_data, 2)

temperature_data
```

```python
weather_df = pd.DataFrame(
    temperature_data,
    index=cities,
    columns=months
)

weather_df
```

```python
annual_average = weather_df.mean(axis=1)

annual_average
```

```python
highest_city = annual_average.idxmax()
lowest_city = annual_average.idxmin()

highest_city
```

```python
lowest_city
```

```python
highest_temperature = annual_average.max()
lowest_temperature = annual_average.min()
```

```python
print(
    f"Highest annual average: {highest_city} "
    f"with {highest_temperature:.2f}°C"
)

print(
    f"Lowest annual average: {lowest_city} "
    f"with {lowest_temperature:.2f}°C"
)
```

```python
weather_df["Annual Average"] = annual_average

weather_df
```

```python
plt.figure(figsize=(12, 7))

for city in cities:
    plt.plot(
        months,
        weather_df.loc[city, months],
        marker="o",
        label=city
    )

plt.title("Monthly Temperature Trends for Each City")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")

plt.legend(
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
annual_average.sort_values().plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Annual Average Temperature by City")
plt.xlabel("City")
plt.ylabel("Average Temperature (°C)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
